## **01: Overview of Conditional Statements**

#### **What is the Need for Conditional Statements?**

Conditional statements are used to make decisions in programming. They allow the code to execute different blocks of code based on certain conditions. This is crucial for creating programs that respond dynamically to different inputs or states.

#### **What Can Conditional Statements Do in Python Programming?**

In Python programming, conditional statements enable:

- **Decision Making:** Execute different code blocks based on conditions.
- **Branching Logic:** Direct the flow of the program based on conditions.
- **Control Structures:** Implement algorithms that require decisions, such as sorting, searching, and more.

Conditional statements are fundamental for building interactive and functional programs that react to various conditions.

## **02: Basic Structure of Conditional Statements**

#### **Simple `IF` Statement**

 The basic syntax of an `if` statement is:

```python
if condition:
    # Operation to be performed when the Condition is True
```

#### **Working of IF Statement**

##### Single Criterion
* if statement can check a single condition and execute a block of code if the condition is `True`
* In this example, the condition `number > 0` is evaluated. If it is `True`, the message "The number is positive." is printed.

```python
number = 10

if number > 0:
    print("The number is positive.")
```

##### Multiple Criteria
* You can also use logical operators to check multiple conditions within a single if statement
* Here, both conditions (`number > 0 and number < 30`) are checked. The and operator ensures that both conditions must be True for the code block to execute.

```python
number = 25

if number > 0 and number < 30:
    print("The number is between 0 and 30.")
```

##### Role of Indentation and Code Blocks
* In Python, indentation is used to define code blocks.
* Code blocks are groups of statements that execute together under the control of a condition.
* Proper indentation indicates which statements belong to the same block.

 Example of correct indentation:

```python
#------Statements 1 and 2 will execute when the `condition` is True
if condition:
    statement1
    statement2
```

##### What happens if proper indentation is not followed?
* If the indentation is not consistent, Python will raise an IndentationError.
* Python relies on indentation to determine the structure of code blocks.

 Example of incorrect indentation:

```python
#------This will cause an IndentationError--------
if condition:
statement1
```

#### **Limitations of Using Just `IF` Statement**
* Using only `if` statements can be limiting because it only handles a single condition.
* For scenarios where multiple mutually exclusive conditions need to be checked, using just if alone is not sufficient.
* More complex logic requires additional structures like `else` and `elif` to handle multiple conditions effectively.

## **03: Knowing More Components of Conditional Statements**

#### **Understanding the `ELSE` Statement**

The `else` statement is used in conjunction with `if` statements to provide an alternative block of code to execute when the `if` condition evaluates to `False`. This ensures that there is always a fallback action, making the logic more robust.

**The `else` statement brings the following additional features:**
- Default Action: It handles all cases where `if` conditions is not met, thus providing a default action.
- Simplifies Code: Reduces the need for complex nested `if` statements by catching all remaining cases with a single block of code.

**Example of Using `ELSE`**
* In this example, if the condition number > 0 is False, the `else` block executes, printing "The number is not positive."

```python
number = -5

if number > 0:
    print("The number is positive.")
else:
    print("The number is not positive.")
```

#### **Explaining the Code Flow with `ELSE`**
The flow of execution when using `else` is straightforward:

* **Evaluate Condition**: The `if` condition is checked.
* **Execute if Block**: If the condition is `True`, the code block under `if` executes.
* **Execute else Block**: If the condition is `False`, the code block under `else` executes.

**Example Code Flow**

```python
number = 7

if number % 2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")
```

#### **Pitfalls of using `ELSE`**

The `else` statement is a catch-all for any conditions not explicitly handled by preceding `if` statements. This can sometimes lead to overgeneralization, where the `else` block might not distinguish between various cases that should be treated differently.

##### **Example: Categorizing Age Groups**

*Consider a scenario where we want to categorize people into different age groups: Child, Teen, Adult, and Senior. Using only `if` and `else` can lead to a lack of clarity and specificity in categorization.*

**Using else in this case:**

* **Overgeneralization**: The else block captures all ages 13 and older without specifying which age group they fall into (Teen, Adult, or Senior). This results in a broad classification and lacks detail.
* **Lack of Specificity**: The code does not handle different age ranges between 13 and 65 (Teen, Adult, Senior) separately, making it less useful for specific age-related tasks or analyses.

```python
age = 25

if age < 13:
    print("Child")
else:
    print("Teen or Adult or Senior")
```

#### **Limitations of using just `IF` and `ELSE`**
* Using only `if` and `else` can be limiting when dealing with multiple mutually exclusive conditions.
* In such cases, it is challenging to manage complex decision-making processes, as `else` can only handle all unspecified conditions without differentiating between them.
* For more granular control, `elif` (else if) statements are needed.

## **04: Handling More Than Two Scenarios with Conditional Statements**

#### **Understanding the `ELSE IF` or `ELIF` Statement**

* The `elif` statement, short for "else if," is used to handle multiple conditions in a more organized way.
* It allows you to check multiple conditions sequentially and execute different blocks of code based on which condition evaluates to `True`.

**Example : Consider a scenario where we want to determine the grade based on a score**

In this example:
* The `if` statement first checks if the score is 90 or higher. If this condition is `True`, it prints "Grade: A"
* If the score is not 90 or higher, the `elif` statement checks if the score is 80 or higher. If this condition is `True`, it prints "Grade: B"
* This process continues with subsequent `elif` statements for other ranges
* If none of the conditions are met, the `else` block is executed, printing "Grade: F"

```python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
else:
    print("Grade: F")
```

#### **General Code Flow in Conditional Statements using IF, ELIF and ELSE**
* First, the program checks the condition in the `if` statement.
* If the condition is `False`, it moves to the `elif` statements in the order they are written, checking each condition until one evaluates to True.
* Finally, if none of the `if` or `elif` conditions are True, the `else` block is executed.

#### **Benefits of Using ELIF**
* **Organizes Multiple Conditions**: `elif` allows you to manage and check multiple conditions in a clear, linear fashion.
* **Improves Readability**: It avoids deep nesting of `if` statements and makes the code more readable and maintainable.
* **Handles Complex Scenarios**: It provides a structured way to handle complex scenarios where multiple conditions need to be evaluated.

Using `elif` effectively helps in creating more flexible and readable conditional logic in your Python programs.

## **05: Using Nested Conditional Statements**

#### **Introduction to Nested Conditional Statements**

Nested conditional statements refer to the use of `if`, `elif`, and `else` statements within other `if`, `elif`, or `else` blocks. This allows for more complex decision-making processes by evaluating conditions at multiple levels.

#### **Real-World Use Case**

* Consider a scenario where you need to determine the appropriate discount for a customer based on their membership level and total purchase amount:
* Explanation
 * First Level: The outer `if` statement checks the membership_level.
 * Second Level: Within each membership level block, the nested `if` statements check the total_purchase amount to determine the appropriate discount.
 * Else: If none of the conditions in the outer `if` or `elif` blocks match, the `else` block provides a default discount.

```python
membership_level = 'Gold'
total_purchase = 1200

if membership_level == 'Gold':
    if total_purchase > 1000:
        discount = 0.20
    elif total_purchase > 500:
        discount = 0.10
    else:
        discount = 0.05
elif membership_level == 'Silver':
    if total_purchase > 1000:
        discount = 0.15
    elif total_purchase > 500:
        discount = 0.07
    else:
        discount = 0.03
else:
    discount = 0.00

print(f"Discount: {discount*100}%")
```

#### Tips for Understanding Nesting and Indentation
Indentation: Ensure that each nested if, elif, and else block is indented correctly. Python uses indentation to define the scope of code blocks.

 Correct indentation example:

```python
if condition1:
    if condition2:
        # Code block when condition2 is True
    else:
        # Code block when condition2 is False
```

#### **Benefits of Nested Conditional Statements**
* **Complex Decision Making**: Allows for more detailed and granular decision-making processes by handling multiple levels of conditions.
* **Flexibility**: Provides flexibility to handle various scenarios and combinations of conditions.
