# **01 Understanding the `range()` Function in Python**

In programming, there are often situations where we need to work with a **sequence of Integers** (0,1,2,3....), such as counting items, iterating over a set of values, or generating a range of dates.
* The `range()` function in Python is a powerful tool that helps us create these sequences of Integers easily without having to write everything by hands
* It is commonly used in Python to run For loops a certain number of times or to perform a task a specific number of times
* It is particularly useful to iterate over a sequence of Integers (e.g., 10, 20, 30, ... or 100, 90, 80, ...)

# **02 Basic working of `range()` function**
* The simplest way to use the `range()` function is by passing a single argument, which represents the end value of the **Integer sequence**
* The `range(n)` function will generate a sequence of numbers **starting from 0** and ending at n-1, i.e., ***the value before the specified number***


```python
print(range(5))
```

```python
type(range(5))
```

#### **Explanation**
* The `range()` function generates a range object, which is a special data type in Python used to represent a sequence of Integers
* This range object stores multiple Integer values but they are not displayed when printed

##### **Displaying the entire sequence of numbers**
* To display all the values generated by the `range()` function, we can convert the range object to a list using the `list()` function
* The `list()` function is a built-in function in Python that takes a sequence and creates a list containing all the elements from the sequence. We will deep-dive more into Lists in the later sections

```python
print(list(range(5)))
```

#### **Why range(5) gives 0,1,2,3,4 and not 0,1,2,3,4,5**
* In Python's counting (indexing) starts from 0
* This means that the first element in a sequence (like a list or a range) is at index 0, the second element is at index 1, and so on. The last index is always one less than the length of the sequence.
* When you call `range(5)`, the `range()` function generates a sequence of numbers starting from 0 and ending at 4 (the value before 5).
* So, the sequence generated by `range(5)` represents the indices from 0 to 4, which correspond to the numbers 0, 1, 2, 3, and 4.
* This concept will get more clear when Indexing is formally introduced in the later sections

# **03 Deep-diving into `range()` function**

#### **Specifying the Start and End Values**
* You can also specify a different starting value (other than the default 0) by passing two arguments in `range()` function
* Please note that the use of list() function with range() is not mandatory
* list() function in the shown code examples is used only to display the sequence of integers that the range() function generates

```python
print(list(range(2, 7)))
```

```python
print(list(range(100, 111)))
```

#### **Understanding the default 'step size' and changing it as per need**
* By now you would have noticed that the common difference between the generated Integers is 1 (default)
* This common difference is called as 'Step' or 'Step Size' in common convention
* The step size can be tweaked as per need by introducing a third argument in `range()` function

```python
print(list(range(0, 11, 2)))
```

```python
print(list(range(0, 101, 10)))
```

#### **Explaining the three arguments in range(a,b,c)**
* First argument (a) : Defines the start point of the required Integer sequence
* Second argument (b) : Defines the end point of the required Integer sequence (not included in the sequence)
* Third argument (c) : Defines the common difference or 'step size' between the consecutive values in the required Integer sequence

#### **Other variations**

#### Negative Integers can also be generated with `range()` function

```python
print(list(range(-10, 11)))
```

#### A descending sequence can also be generated. Note that the step size for a descending sequence should be a Negative number

```python
print(list(range(50, 39, -1)))
```

```python
print(list(range(50, -1, -5)))
```

# **04 Real-World application examples**

#### **Generating a Sequence of Years**

```python
print(list(range(2010, 2025)))
```

#### **Generating a Sequence of Dates in a month**

```python
end_day = 29
print(list(range(1, end_day+1)))
```

```python
end_day = 31
print(list(range(1, end_day+1)))
```

#### **Generating a Number Table**

```python
num = 7
print(list(range(num,num*10+1,num)))
```
