# **This notebook outlines how Python's `help()` function works and the different contexts in which it can be used**

## **01 Working**
- The primary purpose of the `help()` function in Python is to provide documentation and information about modules, functions, classes, and methods.
- If you are unsure about how to use a specific function or module, you can use the `help()` function followed by the name of the function or module to get detailed information about its usage and parameters.
- A better understanding of the `help()` function allows us to explore and utilize Python's built-in and external libraries effectively.
- It is useful for developers to debug and understand the functionality and parameters of different modules, functions, and methods within their code.


## **02 Use**

#### **Accessing documentation of built-in functions**
* Notice below that round brackets `()` are not mentioned against the `print` function

```python
#-------Accessing the documentation about the print() function and the arguments under it---------
help(print)
```

```python
help(dir)
```

#### **Accessing documentation of built-in Data Structures**
* The output here includes a description of the class, its methods, and how to use them effectively

```python
help(list)
```

```python
help(tuple)
```

#### **Accessing documentation of a specific method in a Data Structure**
* Convention is to write the Data Structure name followed by period (.) and then the method name without round brackets

```python
help(list.append)
```

```python
help(tuple.count)
```

* If you have a list variable, let's say nums = [1,2,3], then help(nums.append) will also work in a similar manner

```python
nums = [1,2,3]
help(nums.append)
```

#### **Interactive Mode**
* Executing `help()` enters an interactive help mode
* **Try inputs like :** `len`, `max`, `min`, `str`, `enumerate` etc. to access their documentation
* Use quit to exit the interactive help mode

```python
help()
```
