## **Python Tuples: Overview and Usage**

In this notebook, we'll explore Python's built-in data structure called Tuple. This will help you understand how Tuples work, especially in comparison to Lists, which we've already covered in our previous lessons.

## **Tuples Overview**

### **What is a Tuple?**

A Tuple is a built-in data structure in Python that is very similar to a List. Both Tuples and Lists are used to store collections of items

### **Key differences with Lists**

- **Immutability**:
  - **Tuples** are immutable, meaning that once a Tuple is created, its elements cannot be changed, added, or removed.
  - This immutability makes Tuples more suitable for fixed collections of items.
  - **Lists**, on the other hand, are mutable, which means their content can be modified after creation.

- **Syntax**:
  - Tuples use parentheses `()` to enclose their elements, while Lists use square brackets `[]`.

### **Similarities with Lists**

Tuples support similar operations as Lists, including:

- **Indexing**: Accessing elements by their position.
- **Slicing**: Extracting a part of the Tuple.
- **Membership**: Checking if an element exists in the Tuple.

Let's look at a few code examples demonstrating these operations on Tuples.


##### Creating a Tuple

```python
my_tuple = (1, 2, 3, 4, 5)
```

##### Indexing

```python
print(my_tuple[2])
```

##### Slicing

```python
print(my_tuple[1:4])
```

##### Membership

```python
3 in my_tuple
```

## **Tuple Methods**
Unlike Lists, Tuples have only two built-in methods:

* `index()`: Returns the index of the first occurrence of a specified value.
* `count()`: Returns the number of times a specified value appears in the Tuple.

Here are examples of these methods:

```python
my_tuple = (1, 2, 3, 1, 2, 1)
```

##### Index() method

```python
print(my_tuple.index(1))
```

##### Count() method

```python
print(my_tuple.count(1))
```

## **Creating Tuples**

#### **Manually Creating Tuples**

**Multiple Elements:**
* Simply separate the elements with commas inside parentheses.

```python
my_tuple = (1, 2, 3, 4)
print(my_tuple)
```

* You can also create Tuples by simply separating the elements with commas **without using parentheses**

```python
my_tuple = 1, 2, 3, 4
print(my_tuple)
```

**Single Element:**
* When creating a Tuple with a single element, you need to include a trailing comma.
* This is because, without the comma, Python might think you're using parentheses to call a function (with one argument).
* The comma helps Python understand that you're creating a Tuple.



```python
single_element_tuple = (1,)
print(single_element_tuple)
```

**Or**

```python
single_element_tuple = 100,
print(single_element_tuple)
```

#### **Creating Tuples from Lists**
Using the tuple() function.

**From a List:**
* Convert a List to a Tuple by passing the List to the tuple() function.

```python
my_list = [1, 2, 3, 4]

my_tuple = tuple(my_list)
print(my_tuple)
```

**From range() function:**
* Convert a range object to a Tuple.

```python
my_tuple = tuple(range(5))
print(my_tuple)
```

## **Tuple Unpacking**

Tuple unpacking is a powerful feature in Python that allows you to extract elements from a Tuple and assign them to individual variables.
* Tuple unpacking involves assigning the elements of a Tuple to separate variables in one statement.
* This can be particularly useful when iterating over sequences of Tuples

Below are a few examples to illustrate how Tuple unpacking works:

#### **Example 01 : Basic Tuple Unpacking**
* In this example, the Tuple coordinates is unpacked into three separate variables: x, y, and z

```python
coordinates = (10, 20, 30)

x, y, z = coordinates

print("x:", x)
print("y:", y)
print("z:", z)
```

* This is infact very similar to **assigning values to multiple variables in a single line**
* Notice that R.H.S in the below assignment operation is actually a tuple (defined without parentheses)

```python
name, age, city = 'Kim', 35, "Tokyo"
```

```python
print(name, age, city)
```

#### **Example 02 : Unpacking in a Loop**
* Tuple unpacking allows us to run For loop over multiple variables simultaneously (and access each element directly)

```python
students = [("John", 85), ("Jane", 92), ("Mike", 78)]
```

* In this loop, the variable `info` takes on each Tuple from the students list
* Here, `info` is a Tuple that holds both the name and the score of a student
* Although we can access the entire Tuple, the elements are not individually extracted.

```python
for info in students:
    print(info)
```

* We use tuple unpacking to directly access each element within the Tuple
* In this loop, name and score are variables that directly receive the individual elements of each Tuple.

```python
for name, score in students:
    print(name,score)
```

#### **Example 03 : Unpacking with zip() function**
* The zip() function combines multiple iterables (like lists) into a sequence of Tuples.
* Tuple unpacking can be used to iterate over these combined elements:
* This example highlights how tuple unpacking enables looping over multiple variables at once, simplifying the process of working with paired data.

```python
names = ["Alice", "Bob", "Charlie"]
scores = [90, 85, 88]

combined = zip(names, scores)

for info in combined:
    print(info)
```

```python
for name, score in combined:
    print(name,score)
```

## **Why Use an Immutable Data Structure Like a Tuple?**
* **Safety:** Immutable data structures like Tuples help prevent accidental modification of data. This can help avoid bugs and ensure that certain data remains consistent throughout the execution of a program.

* **Performance:** Tuples are generally faster than Lists in certain operations due to their fixed size. Since Tuples do not need to support operations that modify their content, they can be optimized for performance in scenarios where data is constant.

* **Memory Efficiency:** Tuples often use less memory compared to Lists because they are immutable and have a fixed size, making them more efficient in scenarios where you need a lightweight and fixed-size collection.

* **Hashability:** Tuples are hashable (provided all their elements are hashable), which means they can be used as keys in dictionaries or elements in sets. Lists, being mutable, cannot be used in these contexts.
