# **Python List Methods**

# **01 - Overview of List Methods**
* The purpose of Data structures is to not only store data/information but also provide ways to manipulate it
* Lists methods are like pre-defined functions that are dedicated to the List Data Structure.
* **They help you to perform certain type of operations on the List(s) at hand**. Examples:
 * Adding or removing items
 * Sorting the items (ascending and descending)
 * Computing the frequency of items in the List
 * Locating the Index of items in the List
 * Other operations

# **02 - Important things to know**
* The convention to use list methods is : `listname.methodname()`
* Notice that the `methodname` is preceded by the `listname` on which they are applied
* One can view all the List methods by using `dir(list)` or `dir(listname)`. Refer the items displayed in the end (without underscores)
* To access the available documentation for these methods run `help(list)` or `help(list.methodname)`
* There are certain methods that modify the list in place versus those that return a new list
* Similar to function arguments, methods too have arguments that can be used to tweak the way method is applied, e.g., using argument in the `list.sort()` method to change sorting to Descending

#### Viewing all List methods
* Refer the items displayed in the end (without underscores)

```python
dir(list)
```

#### Accessing the available documentation for List methods
*  Refer the items displayed in the end (without underscores)

```python
help(list)
```

#### Accessing documentation for a specific List method

```python
help(list.append)
```

# **03 - List Methods**

### **3.1) `append()` method**
- The `append()` method adds a single element to the end of the list.
- It takes one argument, the element to be added.
- It modifies the original list (in place)

##### Example 01:

```python
my_list = [1, 2, 3]
my_list.append(4)
print(my_list)
```

##### Example 02:

```python
books = ["To Kill", "1984", "Pride"]
new = ["Gatsby", "Harry", "Catcher"]
books.append(new)
print(books)
```

### **3.2) insert() method**
- The `insert(index,element)` method inserts an element at a specified position in the list
- It takes two arguments:
 - the `index`, which is the Index position where the element should be inserted, and
 - the `element`, to be inserted
- It modifies the original list (in place)

##### Example 01:

```python
my_list = [10, 20, 40]
my_list.insert(2, 30)
print(my_list)
```

##### Example 02:

```python
grocery_list = ['Apples', 'Bananas', 'Oranges']
grocery_list.insert(2, 'Carrots')
print(grocery_list)
```

### **3.3) extend() method**
- The `extend()` method extends a given list (e.g. `list1`) by appending elements from another list (e.g. `list2`)
- It takes only one argument, which is the name of the list to be appended (e.g. `list2`)
- Here, the elements of `list2` are added to the end of `list1`, thus modifying `list1` (in place)

##### Example 01:

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]
list1.extend(list2)
print(list1)
```

##### Example 02:

```python
grocery_items = ["Milk", "Bread"]
additional_groceries = ["Eggs", "Cheese", "Apples"]
grocery_items.extend(additional_groceries)
print(grocery_items)
```

### **3.4) remove() method**
- The `remove()` method removes the first occurrence of a specified value from the list
- It takes one argument, the value to be removed
- The list is modified (in place) by removing the first occurrence of the specified value
- If the provided value is not found, the method raises a `ValueError`

##### Example 01:

```python
my_list = [1, 2, 3, 4]
my_list.remove(3)
print(my_list)
```

##### Example 02:

```python
pet_names = ['Buddy', 'Bella', 'Buddy', 'Max', 'Buddy', 'Daisy']
pet_names.remove('Buddy')
print(pet_names)
```

### **3.5) pop() method**
- The `pop()` method is used to remove a certain element from a List by specifying its index position
- It takes one optional argument, which is the index of the element to be removed
- If no index is specified, it removes the last element
- When `pop()` method is applied, it does two things:
 - Returns the removed element. This could be saved against a variable
 - Modifies the List (in place) by removing the element at the specified index

##### Example 01:

```python
my_list = [100, 200, 300, 400]
my_list.pop(1)
```

```python
print(my_list)
```

##### Example 02:

```python
colors = ['red', 'green', 'blue']
last_color = colors.pop()
print(last_color)
print(colors)
```

### **3.6) index() method**
- The `index()` method is used to return the index of the first occurrence of a specified value
- It takes one argument, the value to find
- It also has two optional arguments:
 - `start`, which is the starting index for the search (default is 0), and
 - `end`, which is the ending index for the search (default is the end of the list).
- If the searched value is not found, it raises a `ValueError`

##### Example 01:

```python
my_list = [10, 20, 30, 40, 30]
my_list.index(30)
```

##### Example 02:

```python
animals = ['cat', 'dog', 'bird', 'ant','dog','fly']
first_occurrence = animals.index('dog')
print(first_occurrence)
```

```python
# To find the next occurrence of 'dog' we are starting from Index of 2 (i.e., 'bird' onwards)
second_occurrence = animals.index('dog',first_occurrence+1)
print(second_occurrence)
```

### **3.7) count() method**
- The `count()` method is used to find the number of occurrences of a specified value in a list
- It takes only one argument, the value to count
- It returns the number of times (frequency) the specified value appears in the list

##### Example 01:

```python
my_list = [100, 200, 300, 200, 200, 400]
my_list.count(200)
```

##### Example 02:

```python
letters = ['a', 'b', 'A', 'C', 'a']
letters.count('a')
```

### **3.8) sort() method**
- The `sort()` method is used to sort the list items in ascending (default) or descending order
- No arguements need to be passed to sort the List in ascending order
- To sort the List in descending order, use the argument (`reverse` = `True`)
- This method sorts the original list in place

##### Example 01:

```python
nums = [3, 1, 4, 2]
nums.sort()
print(nums)
```

##### Example 02:

```python
fruits = ['banana', 'apple', 'cherry','peach','mango']
fruits.sort(reverse=True)
print(fruits)
```

### **3.9) reverse() method**
- The `reverse()` method is used to reverse the order of elements in a List  
- It does not take any arguments
- It modifies the original list (in place)

##### Example 01:

```python
nums = [1, 2, 3, 4]
nums.reverse()
print(nums)
```

##### Example 02:

```python
collection = [1, 'B', 3, 'D', 5]
collection.reverse()
print(collection)
```

### **3.10) copy() method**
- The `copy()` method returns a shallow copy of a list
- It does not take any arguments
- A new list is created and returned as a shallow copy of the original list

##### Example 01: Making copy of a List without using the List's copy() method

```python
nums = [10, 30, 50, 70, 90]
nums_copy = nums
print(nums)
print(nums_copy)
```

```python
#--Any changes in the List Copy will also reflect in the Original List
nums_copy[0] = 1
print(nums_copy)
print(nums)
```

##### Example 02: Making copy of a List with List's copy() method

```python
numbers = [10, 30, 50, 70, 90]
numbers_copy = numbers.copy()
print(numbers)
print(numbers_copy)
```

```python
#--Changes in the List Copy will NOT reflect in the Original List
numbers_copy[0] = 1
print(numbers_copy)
print(numbers)
```

### **3.11) clear() method**
- The `clear()` method is used to remove all the elements from a list, thus making it blank
- It does not take any arguments
- The original list gets empty (in place)

##### Example 01:

```python
my_list = [1, 2, 3]
my_list.clear()
print(my_list)
```

##### Example 02:

```python
letters = ['a', 'b', 'c']
letters.clear()
print(letters)
```
