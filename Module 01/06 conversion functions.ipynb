# **This notebook describes the working of miscellaneous built-in function in Python**
* `int()`
* `float()`
* `str()`

## **01 Working**
* `int()` function is used to convert a value to an integer data type. It can be used to convert a string, float, or other numeric value to an integer
* `float()` function is used to convert a value to a floating-point number (decimal number)
* The `str()` function in Python is used to convert a value to a string. It takes a value as an argument and returns a string. If the input value is not a string, it will convert it to a string

## **02 Use**

#### **Changing to Integer type**

##### Converting a string representation of a number to an integer

```python
num = int("123")
print(num)
type(num)
```

```python
exp_years = int(input("Enter your experience in years:"))
print(exp_years)
type(exp_years)
```

##### Rounding down a float to the nearest integer

```python
num = int(12.113)
print(num)
type(num)
```

```python
num = int(12.993)
print(num)
type(num)
```

##### Converting a Boolean value to its numeric representation

```python
num = int(True)
print(num)
type(num)
```

#### **Changing to Float type**

```python
num = float("123.45")
print(num)
type(num)
```

```python
exchange_rate = float(input("Enter your exchange rate:"))
print(exchange_rate)
type(exchange_rate)
```

```python
num = float(100)
print(num)
type(num)
```

```python
num = float(False)
print(num)
type(num)
```

#### **Changing to String type**

```python
num = str(123)
type(num)
num
```

```python
num = str(123.45)
type(num)
num
```

```python
num = str(False)
type(num)
num
```

## **03 Note on these conversion functions**
* The int(), float(), and str() functions are essential tools in Python programming, allowing you to convert values between different data types. These functions are crucial for ensuring data consistency, enabling proper operations, and facilitating effective data processing and manipulation.
* Use at different levels of Python Proficiency :
 * **Beginner Level:** At the beginner stage, these conversion functions are essential for handling user input and performing basic data processing tasks. Learners quickly realize the need to convert input strings to numeric types (using int() and float()) to perform calculations or comparisons.
 * **Intermediate Level:** As learners delve into more complex data structures, such as lists, dictionaries, and custom classes, the conversion functions become crucial for maintaining data integrity and enabling seamless operations. Learners use these functions to ensure data consistency and facilitate operations like sorting, filtering, and aggregating.
 * **Advanced Level:** At the advanced level, learners leverage the conversion functions in conjunction with control structures (e.g., conditional statements, loops), functions, and modules to build sophisticated, dynamic, and user-friendly applications. The ability to convert data types becomes essential for error handling, data validation, and implementing complex business logic.
