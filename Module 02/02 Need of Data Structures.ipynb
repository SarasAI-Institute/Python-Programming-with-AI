### **Need of a Data Structure**
- Data structures help you organize and manage data efficiently.
- They make it easier to store, access, and modify data.
- Using the right data structure can optimize your code and improve performance.
- They are essential for handling large amounts of data and complex operations.

#### **The following example will help us understand the limitations of using simple variables and also the need of having a Data Container (Data Structure) that can store multiple values**
* Example : Let's say we want to observe "How the Interest rates offerred by a Bank on Auto Loans have changed over time?"
* To store this information of Interest rates offerred for e.g., last four quarters (in the last Calendar year) we will need to create Four variables

```python
Q1 = 0.07
Q2 = 0.075
Q3 = 0.08
Q4 = 0.077
```

 * To store the same information for the last Eight quarters (across last Two Calendar years) we will need to create Eight variables
 * Notice not only we need more variables, but we also need to be creative in coming up with appropriate variable names that correctly indicate the time periods for these values

```python
Q1_2022 = 0.06
Q2_2022 = 0.065
Q3_2022 = 0.072
Q4_2022 = 0.070

Q1_2023 = 0.07
Q2_2023 = 0.075
Q3_2023 = 0.08
Q4_2023 = 0.077
```

* ***Imagine the number of variables we need to create (and their names) if this exercise was to be done for the Last five years and for Ten different Banks***

#### **Let's look at another aspect, i.e., Performing operations with these multiple variables**
* Example : Let's say we want to compute the average interest rates offered in the last calendar year (across the four quarters)
* A simple expression involving the Four variables (stored earlier) should help get the average interest rate value

```python
(Q1_2023 + Q2_2023 + Q3_2023 + Q4_2023)/4
```

* We need an even more sophisticated expression to compute average interest rate value observed in the last Two calendar years (across the Eight quarters)

```python
(Q1_2023 + Q2_2023 + Q3_2023 + Q4_2023 + Q1_2022 + Q2_2022 + Q3_2022 + Q4_2022)/8
```

* ***Imagine writing the expression to compute average Interest Rate in the Last five years***
* **Clearly this approach is NOT SCALABLE !!**

## **Summarizing our learnings from the example**
* The primary need is to store a collection of related values in a single, organized data structure without the burden of creating and managing multiple individual variables. This approach allows for more concise and manageable code
* Accessing the individual elements within data structure should be straightforward, enabling efficient retrieval and manipulation of the stored values (to be explained in later sections of this Module)
* Performing operations on the multiple values stored in the data structure should not require writing complex or cumbersome expressions. The data structure should facilitate intuitive and efficient operations, simplifying the overall logic and implementation

#### **Actually, We can do a lot more !!**

##### len() function could be used to count the number of values or items in the List.
* This way we can write the "Average Interest Rate" expression where the denominator is not fed manually.
* In future if we decide to add more Interest Rate values to our list (e.g. last Three years), the len() function will automatically detect the length of the updated list, thus avoiding any manual tweaks in the expression

```python
len(int_rates_1y)
```

```python
sum(int_rates_1y)/len(int_rates_1y)
```

```python
sum(int_rates_2y)/len(int_rates_2y)
```

#### Using the max() function to compute the highest value of Interest Rate

```python
max(int_rates_2y)
```

#### Using the min() function to compute the lowest value of Interest Rate

```python
min(int_rates_2y)
```
