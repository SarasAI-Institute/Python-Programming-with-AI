# **This notebook outlines how Python's `type()` function works and the different contexts in which it can be used**

## **01 Working**
* The primary purpose of the `type()` function in Python is to determine and return the type of a given object.
* If you are unsure whether a given variable is integer, float, boolean, string or let's say a Data Structure (`List`, `Tuple`, etc.), you can use the `type()` function on the variable name
* A better knowledge of the 'type of the object' helps us apply appropriate operations on it
* It is useful to debug and understand the behavior of different variables and expressions

## **02 Use**

#### **Determining Data Types of values**

```python
type('Hello World')
```

```python
type('1000')
```

```python
type(100)
```

```python
type(True)
```

```python
type(4.56)
```

```python
type(10+9j)
```

#### **Determining Data Types expression outputs**

```python
print(99 + True)
type(99 + True)
```

```python
print(100/10)
type(100/10)
```

```python
print(7 + 3.5)
type(7 + 3.5)
```

#### **Comparing `type()` function output against a criteria**
* **Comparisons like below can be valuable for executing operations based on conditions.**
* **Note :** These comparisons are particularly useful for conditional operations, which will be explored further in the later modules

```python
type(100) == int
```

```python
type(3.45) == int
```

```python
type(10 + 2.9) == float
```

```python
type(10>7) == bool
```

```python
type(10!=7) == bool
```

```python
type('Hi'*2) == str
```
