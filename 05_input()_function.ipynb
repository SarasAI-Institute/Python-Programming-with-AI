# **This notebook outlines how Python's `input()` function works and the different contexts in which it can be used**

## **01 Working**
* The primary purpose of the `input()` function in Python is to accept user input from the keyboard or console
* When the `input()` function is called, the program execution pauses and waits for the user to enter some data and press Enter
* By default user's input is captured as a string, even if the user enters numbers or other characters

#### A basic example
* *Execute the below code and enter any kind of input (number, name, statement) and observe the output*

```python
input()
```

## **02 Practical Uses**
The `input()` function is commonly used to:
* Get information from the user during program execution
* Create interactive programs that respond to user input
* Implement simple command-line interfaces

Overall, the `input()` function is a powerful tool for making Python programs more interactive and user-friendly

#### **Prompting the user to enter a value and storing it in a variable**
* Here we are asking the user to provide their name
* The user inputted value gets stored against a variable 'username'

```python
username = input("What is your name? ")
print(f"Hello, {username}!")
```

#### **Input with Type Conversion**
* Example 1 : Here the user input is converted to an Integer by using the `int()` function, which is another built-in function

```python
age = int(input("Enter Age :"))
print(f"I am {age} years old!")
```

* Example 2 : Here the user input is converted to a float value by using the `float()` function, which is another built-in function

```python
interest_rate = float(input("Enter loan's interest rate :"))
print(f"My loan has an interest rate of {interest_rate} ")
```

#### **Taking multiple inputs from user**

```python
name = input("Enter your Name :")
age = int(input("Enter your Age :"))
height = float(input("Height in Inches :"))
print()
print(f"Name : {name}",f"Age : {age}",f"Height : {height}",sep='\n')
```
