# **This notebook outlines how Python's `print()` function works and the different contexts in which it can be used**

## **01 Working**
* The main purpose of the `print()` function in Python is to output data to the
console.
* It is commonly used to display messages, variable values, and results of computations, making it a fundamental tool for debugging and providing user feedback in a program

## **02 Basic Use**

#### **Displaying Text**

```python
print('Hello World')
```

#### **Displaying output of expressions**

```python
print(2+3+4)
```

```python
print(3*(10-7)+1)
```

#### **Displaying multiple outputs**

```python
#-----without print() function only the last evaluated expression output is displayed
a = 100
b = 20
a+b
a-b
```

```python
#-----with print() function we can display the outputs of multiple expressions
a = 100
b = 20
print(a+b)
print(a-b)
```

## **03 Advanced Use**
* Adding multiple arguments to the `print()` function, separated by commas, enhances output readability by automatically inserting spaces between the arguments.
* This allows you to combine different types of data (e.g., strings, numbers, variables) in a single print statement without manually adding spaces or converting data types

#### **Adding strings in the `print()` function to describe the displayed ouput**

```python
a = 100
b = 20
print('total', a+b)
print('difference', a-b)
```

#### **Adding more strings and variables to improve the clarity and interpretability of the output**

```python
a = 100
b = 20
print('Sum of', a ,'and', b ,'is :', a+b)
print('Difference of', a ,'and', b ,'is :', a-b)
```

#### **Using f string in `print()` function**
* An f-string is a type of string that starts with the letter f or F and allows you to embed variables and expressions directly within curly braces `{}`.
* This approach enables the dynamic insertion of variables and expressions into a string, enhancing readability and making the code more concise compared to using multiple arguments in the `print()` function.

```python
#-----------Example 01-----------
name = 'John Smith'
print(F'My name is {name}')
```

```python
#-----------Example 02-----------
name = 'John Smith'
age = 38
print(F"Hello, my name is {name} and I am {age} years old.")
```

```python
#-----------Example 03-----------
a = 100
b = 20
print(f'Product of {a} and {b} is : {a*b}')
print(f'Ratio of {a} to {b} is : {a/b}')
```

## **04 Important arguments under the `print()` function**
* **`sep` argument**
 * Specifies the string that will be inserted between multiple arguments in the `print()` function
 * The default value of this argument is a single space `' '`
 * It's useful when you want to format the output with a specific delimiter, such as a comma, hyphen, or no space at all
* **`end` argument**
 * Specifies the string that will be displayed at the end of the printed output
 * The default value is a newline character `'\n'`, which moves the cursor to the next line
 * It's useful when you want to continue printing on the same line or use a different ending character, such as a tab space `'\t'` a single space or a custom message

#### **Separator argument (`sep`) in `print()` function**

```python
print('A','B','C',sep="|")
```

```python
print(1,2,3,4,5,sep="@")
```

```python
nation_code = '+1'
area_code = '234'
phone_num = '567890'
print(nation_code,area_code,phone_num,sep='-')
```

#### **End argument in `print()` function**
* Default value is the `'\n'` character, which is the reason outputs of multiple `print()` statements show in new lines

```python
#-----------Example 01-----------
print('A','B','C')
print('D','E','F')
print(1,2,3)
```

* Passing the argument value as Tab space character `'\t'`

```python
print('A','B','C',end="\t")
print('D','E','F',end="\t")
print(1,2,3)
```

```python
#-----------Example 02-----------
name = 'John Smith'
age = 34
gender = 'Male'
city = 'Cairo'
print('Name :',name)
print('Age :', age)
print('Gender :',gender)
print('City :', city)
```

```python
name = 'John Smith'
age = 34
gender = 'Male'
city = 'Cairo'
print('Name :',name,end="\t")
print('Age :', age)
print('Gender :',gender, end="\t")
print('City :', city)
```
