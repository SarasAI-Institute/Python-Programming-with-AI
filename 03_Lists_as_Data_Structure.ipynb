- **Thus, irrespective of the List length using the Index value of -1 will always return the last item from the List**

### **04 Performing Membership operations on a List**
- We can check if our List contains a certain value - e.g. checking if our list 'movies' has the movie name 'Catch'
- This is done by using the `in` operator - e.g. `'Catch' in movies`
- It returns `True` if the value is found in the list, otherwise it returns `False`
- Similarly we can use the `not in` operator, which returns `True` if the value we are looking for is not found in the list, otherwise it returns `False`
- Both `in` and `not in` are called **membership operators** since they check if a given value is a member of a Data Structure, e.g., List

```python
movies
```

```python
'Catch' in movies
```

```python
'True Lies' not in movies
```

### **05 Slicing on a List**

**The Concept**
- Earlier we learned how to access individual items from a List using Indexing
- There are many situations where we need to fetch a range of items from a List - for example:
 - The first three items
 - Items from index position 4 up to 8
 - The last five items
- Notice that in all these examples, the items we're accessing are contiguous, meaning they are positioned next to each other in the List
- Similarly we may be required to access items at specific non-contiguous index positions following a certain pattern. For example :
 - All items at even positions (index values 0, 2, 4, 6, ...)
 - All items at odd positions, (index values 1, 3, 5, 7, ...)
 - All items at positions with a multiple of 5 (index values 0, 5, 10, 15, ...)
- **Slicing can be used to easily access these multiple elements from a List**

**The Execution**
- To perform Slicing on a List we again use Square Brackets [ ] after List name, but unlike Indexing, here we pass a range of Index values
- This range of Index values is declared by using a `Start` and `End` point and placing a colon `:` in between. **Here `Start` is inclusive and `End` is exclusive**
 - Example 01 : **`listname[0:4]`** will provide us the List items with Index values 0 and 3
- Other than `Start` and `End` points we can add a third value after another colon `:`, which represents the step size
 - Example 02 : **`listname[0:10:2]`** will provide us the List items with Index values 0, 2, 4, 6 and 8
 - **`listname[0:10]`** would normally return the elements at indices 0 through 9, but with the step size of 2 it is different
 - Python will start at the first element (index 0), then move to the element at index 2, then to index 4, and so on, skipping every other element until it reaches the element at index 9. This results in a list containing the elements at indices 0, 2, 4, 6, and 8.
 - In summary, the `Step` value in slicing allows you to specify a pattern for selecting elements, such as every other element, every third element, and so on
- **Note that Slicing operation returns a List as the output**

```python
list(range(10))[0:4]
```

```python
list(range(10))[0:10:2]
```

#### **More variations in Slicing**

1. **Not mentioning the start point - `[ : 5]`**
 - When the start point is not mentioned, it defaults to 0, so this is equivalent to [0:5].
 - It returns the elements from the beginning of the sequence up to, but not including, index 5.
2. **Not mentioning the end point - `[5 : ]`**
 - When the end point is not mentioned, it defaults to the length of the sequence, so this is equivalent to [5:len(sequence)].
 - It returns the elements from index 5 up to the end of the sequence.
3. **Mentioning both start and end points but not step - `[1 : 10 : ]`**
 - This is equivalent to [1:10], as the step size defaults to 1 when not provided.
 - It returns the elements from index 1 up to, but not including, index 10.
4. **Not mentioning the end point but step -  `[5 : : 2]`**
 - When the end point is not mentioned, it defaults to the length of the sequence.
 - The step size is set to 2, so this returns every other element starting from index 5 until the end of the sequence.
5. **Not mentioning anything - `[ : : ]`**
 - This is a shorthand for [0:len(sequence):1], which returns a copy of the entire sequence.
6. **Mentioning a negative step size - `[ : : -1]`**
 - The negative step size of -1 reverses the order of the elements in the sequence.
 - This is a common way to reverse a list or string.
7. **Reverse slicing on a list with 10 items with step size of 3 - `[ -7 : : -3]`**
 - The start index -7 corresponds to index 3 from the end of the list (10 - 7 = 3).
 - The step size of -3 means it will select every third element in reverse order, starting from index 3.
 - This will return the elements at indices 3, 0 from the end of the list.
