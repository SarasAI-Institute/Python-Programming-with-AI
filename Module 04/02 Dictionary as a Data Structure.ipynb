# **Section 03 : How to Access and Update Dictionary Values**

* Now that we understand the structure of dictionaries, let’s explore how you can use keys to access and update the values stored in them. This method is known as **key-based indexing**. It’s one of the most powerful features of dictionaries, allowing you to directly interact with your data.

### **What is key-based indexing?**
  - Key-based indexing means using the **key** in a dictionary to directly access its associated **value**.
  - Unlike lists or tuples, **dictionary items are not ordered**. This means that dictionary items do not have integer indices, which you would use in a list or tuple.
  - Instead, each value in a dictionary is accessed by its corresponding key, making key-based indexing more relevant and practical for retrieving and updating information.

### **Answering the earlier posed questions using key-based indexing:**

  1. **Finding the Grade for English subject:**  
   - Use the key `'English'` to access the grade.

```python
grades_dict = {'Math': 85, 'English': 90, 'Science': 92}
print(grades_dict['English'])
```

*   **Explanation:**
    - Here, `'English'` is the key, and `90` is the value. By using the key, you directly get the grade without worrying about its position.

2. **Updating the Quantity of grocery item Milk:**
   - Use the key `'Milk'` to change its value.

```python
grocery_dict = {'Apples': 4, 'Bananas': 6, 'Milk': 1}
grocery_dict['Milk'] = 2
print(grocery_dict)
```

- **Explanation:**
 - By using `'Milk'` as the key, you can locate the quantity and update it directly. The new quantity is now `2`.

3. **Retrieving Phone Number and updating Email:**

* **3.1 Retrieving Charlie's Phone Number:**

  **Access Charlie's information**
   - Use the key `'Charlie'` to access his contact details in the `friends_dict` dictionary.
  
  **Retrieve the Phone Number from the inner dictionary:**
   - From the inner (nested) dictionary, use the key `'phone'` to get Charlie's phone number.
   - Note `charlie_details` is the inner dictionary where Charlie's contact details (Phone number and Email ID) are stored

```python
charlie_details = friends_dict['Charlie']

charlie_phone = charlie_details['phone']
```

```python
# or directly
charlie_phone = friends_dict['Charlie']['phone']
```

 * **3.2 Updating Charlie's Email Address:**
    
    **Access Charlie's Contact Details:**

  **Update the Email Address:**

      - Also overwrite it by using the assignment operator and declaring a new email address on RHS



```python
charlie_details = friends_dict['Charlie']
charlie_details['email'] = 'newcharlie@example.com'
```

```python
# or directly
friends_dict['Charlie']['email'] = 'newcharlie@example.com'
```

- **Why It Matters:**
  - Key-based indexing makes retrieving and updating data straightforward and efficient. You don’t have to search through a list or remember the exact position of an item—just use the key, and you get what you need directly.

# **Section 04 : Looping Through Dictionaries**

### **Overview**
Dictionaries in Python are powerful data structures that store data in key-value pairs. Often, you need to process or access this data efficiently. Loops are essential for this purpose, allowing you to iterate over the dictionary’s contents.

**Why Use Loops with Dictionaries?**

- **Data Processing:** Loops help process or manipulate each key-value pair in a dictionary, such as calculating totals or generating reports.
- **Data Access:** You might need to retrieve specific pieces of data or perform actions based on dictionary contents.
- **Automation:** Loops automate repetitive tasks, reducing the need for manual data handling and improving code efficiency.

Python provides several methods to loop through dictionaries: `.items()`, `.keys()`, and `.values()`. Each method offers different advantages depending on whether you need to access keys, values, or both.

### **Looping through Dictionary Items using the `.items()` method**

**Using `.items()` Method:**
   - The `.items()` method returns a view of the dictionary’s key-value pairs as tuples.
   - Each tuple consists of a key and its associated value.

```python
grades_dict = {'Math': 85, 'English': 90, 'Science': 92}
items = grades_dict.items()
print(items)
```

```python
for info in grades_dict.items():
  print(info)
```

**Fetching Keys and Values via Tuple Unpacking:**
   - You can use a `for` loop to iterate through these tuples.
   - Each tuple is unpacked into its key and value.

```python
for subject, grade in grades_dict.items():
    print(f"Subject: {subject}, Grade: {grade}")
```

**Explanation:**

- **Access Key-Value Pairs:** The `.items()` method provides access to all key-value pairs in the dictionary.
- **Tuple Unpacking:** The method returns each pair as a tuple. You can use tuple unpacking in the loop to access both the key and the value directly.

- **Best Practice:** Use `.items()` when you need to work with both keys and values together, such as when displaying or processing both parts of each entry.

### **Looping through Dictionary Keys using the `.keys()` method**

**Using `.keys()` Method:**
   - The `.keys()` method returns a view of the dictionary’s keys.

```python
grades_dict = {'Math': 85, 'English': 90, 'Science': 92}
keys = grades_dict.keys()
print(keys)
```

**Looping through Keys:**
   - You can use a `for` loop to iterate through these `keys`.

```python
for subject in grades_dict.keys():
    print(subject)
```

- Iterating directly over the dictionary without using `.keys()` will give the same result as `.keys()`

```python
for subject in grades_dict:
  print(subject)
```

**Best Practice:**
- Use `.keys()` when you need to work with or display only the keys, or when you need to perform operations that involve keys without needing their values.

### **Looping through Dictionary values using the `.values()` method**

**Using `.values()` Method:**
   - The `.values()` method returns a view of the dictionary’s values.



```python
grades_dict = {'Math': 85, 'English': 90, 'Science': 92}
values = grades_dict.values()
print(values)
```

**Looping through Values:**
   - You can use a `for` loop to iterate through these `values`.

```python
for grade in grades_dict.values():
    print(grade)
```

**Best Practice:**
- Use `.values()` when you need to work with or display only the values, or when performing calculations or operations that involve only the values.

### **Examples for further understanding looping in Dictionaries**

##### **Example 1: Student Grades Dictionary**

   - A nested dictionary where each student has a dictionary of subjects and grades:

```python
students_grades = {
    'Alice': {'Math': 88, 'English': 92, 'Science': 85},
    'Bob': {'Math': 79, 'English': 81, 'Science': 78},
    'Charlie': {'Math': 95, 'English': 90, 'Science': 94}}
```

```python
# Loop through students and their grades.
for student, grades in students_grades.items():
    print(f"Student: {student}")
    for subject, grade in grades.items():
        print(f"  Subject: {subject}, Grade: {grade}")
    print()
```

**Example 2: Product Inventory Dictionary**
- A dictionary where each product has a price and stock quantity:

```python
product_inventory = {
    'Laptop': {'price': 1200, 'stock': 30},
    'Smartphone': {'price': 800, 'stock': 50},
    'Headphones': {'price': 150, 'stock': 100}}
```

```python
# Loop through product names and look up their prices and stock quantities.

for product in product_inventory.keys():
  price = product_inventory[product]['price']
  stock = product_inventory[product]['stock']
  print(f"Product: {product}, Price: ${price}, Stock: {stock}")
```

**Example 3: Company Employee Salaries**
- A dictionary where each department has employees with their salaries:

```python
company_salaries = {
    'HR': {'Alice': 70000, 'Bob': 75000},
    'Engineering': {'Charlie': 90000, 'David': 95000},
    'Marketing': {'Eve': 65000, 'Frank': 67000}}
```

# **Section 05 : Creating Dictionaries**

### **Manually**

* **Starting with an Empty Dictionary:**
   - Imagine you’re setting up a new dictionary where you want to store some information.
   - You start with an empty dictionary and then begin adding the information you need.

```python
my_dict = {}

my_dict['name'] = 'Alice'
my_dict['age'] = 30
my_dict['city'] = 'New York'

print(my_dict)
```

   - **Explanation:**
     - Here, we start with an empty dictionary `my_dict` and add items to it.
     - Each item is a key-value pair, where `'name'`, `'age'`, and `'city'` are **`keys`** and `'Alice'`, `30`, and `'New York'` are their respective **`values`**.

* **Updating Information:**
   - If you need to update some information, like changing Alice's age, you can do so easily. Just update the value associated with the existing key.

```python
my_dict['age'] = 31

print(my_dict)
```

   - **Explanation:**
     - In this example, we updated Alice's age from 30 to 31.
     - This shows how flexible dictionaries are, you can change values without affecting the keys.

* **Handling Key Uniqueness:**
   - **Remember, each key in a dictionary must be unique.**
   - If you try to add a new value for an existing key, the old value will be replaced.

```python
my_dict['city'] = 'Los Angeles'

print(my_dict)
```

   - **Explanation:**
     - Here, we replaced the old value `'New York'` with `'Los Angeles'` for the key `'city'`.
     - Dictionaries automatically handle this for you, so you don’t have to worry about duplicates.

### **With dict() and zip() functions**

* **Start with Simple Lists:**
   - Imagine you have a list of students and a corresponding list of their scores. You can combine these lists into a dictionary.

```python
students = ['James', 'Sophie', 'Alex']
scores = [88, 92, 85]
```

* **Combine Lists into a Dictionary:**
   - Use `zip()` to pair each student with their score. Then, use `dict()` to turn these pairs into a dictionary.

```python
student_scores = dict(zip(students, scores))

print(student_scores)
```

* **Explanation:**
     - `zip()` pairs up each student with their score, creating tuples like `('James', 88)`.
     - The `dict()` function then converts these tuples into a dictionary.
     - This method is quick and efficient for turning two related lists into a dictionary.

* **When to Use:**
 - **`zip()` and `dict()`:** Perfect for quickly combining two lists into a dictionary when you have the lists of `keys` and `values` ready.

### **Iteratively - Using a For loop**

* **Start with simple collections : e.g., Lists:**
   - Imagine you have a list of products and a list of their prices. We’ll use these lists to make a dictionary that links each product to its price.

```python
products = ['Laptop', 'Smartphone', 'Headphones']
prices = [1200, 800, 150]
```

* **Combine Lists using `zip()`:**
   - The `zip()` function pairs up each product with its price.

```python
zipped = zip(products, prices)
print(list(zipped))
```

**Explanation:**
- The `zip()` function combined the items from the two lists into pairs. Each pair is a tuple that contains one element from each list.

- For example, `zip(products, prices)` pairs each product with its price, creating tuples like `('Laptop', 1200)`, `('Smartphone', 800)`.... these tuples are then used to build the dictionary, where:
    - `'Laptop'` becomes the key.
    - `1200` becomes the value.

- Without the `zip()` function, you would need to access the items from the lists at the corresponding Index positions, which requires writing the for loop on Index values (a longer method to get the same output)
- `zip()` simplifies this process by handling the pairing for you, making it easy to create a dictionary from two related lists.

* **Creating a Dictionary Using for Loop:**
   - We’ll use a for loop to add each pair to a dictionary. This way, each product will be linked to its price.

```python
product_dict = {}

for product, price in zip(products, prices):
    product_dict[product] = price

print(product_dict)
```

**Explanation:**
- **Code Line: `product_dict[product] = price`**
  - **What It Does:** This line puts each product and its price into the dictionary (as an item).
  - **How It Works:**
    - `product_dict[product]`: This means we are adding a new item where `product` is the key. For example, if `product` is `'Laptop'`, this part refers to `product_dict['Laptop']`.
    - `= price`: This sets the value for that key. So, `product_dict['Laptop']` will be `1200`.
  - This helps you create a dictionary that connects each product to its price.
  - If you try to add a new price for an existing product, it will update the old price with the new one.

### **Practical example to understand the difference between the above Two methods**

**Example : Creating employee records with For Loops**
- You have employee data in three separate lists: `names`, `salaries` and `departments`

- You want to create a dictionary where:
  - Each employee's name is the key.
  - The value is another dictionary with their salary and department.
  - Apply a 10% bonus to salaries for employees in the 'Sales' department.

```python
names = ['Alice', 'Bob', 'Charlie']
salaries = [50000, 60000, 55000]
departments = ['HR', 'Sales', 'Sales']

employee_dict = {}
bonus_percentage = 0.10
for name, salary, department in zip(names, salaries, departments):
    if department == 'Sales':
        salary += salary * bonus_percentage
    employee_dict[name] = {'Salary': salary, 'Department': department}

employee_dict
```

**Explanation:**

* **Data Processing:** The for loop allows you to apply a specific condition (bonus for 'Sales' employees) directly during dictionary creation. This is not feasible with `dict(zip())`, as it does not handle such conditional logic.
* **Flexibility:** Using for loops provides greater flexibility for additional data processing or transformations during dictionary creation, making it suitable for more complex scenarios.

### **Which method to use for Dictionary creation :**

 * For Loops: Ideal when you need to perform additional operations or conditions during dictionary creation, such as applying transformations, filtering data, or merging multiple sources with specific rules.

 * zip() and dict(): Best for straightforward cases where lists of `keys` and `values` are combined directly without extra processing.

# **Section 06 : Dictionary Methods**
* Dictionaries in Python come with several methods that make it easier to manipulate and retrieve data. These methods are designed to handle common tasks such as retrieving values, adding or removing key-value pairs, and updating existing entries. Understanding these methods can significantly enhance your ability to work with dictionaries effectively.

### **`get()` method**
- The `get()` method helps you find the value for a specific key in a dictionary. If the key isn’t there, it won’t cause an error; instead, you can choose what message or value to show.

```python
student_scores = {'Math': 90, 'English': 85, 'Science': 92}

math_score = student_scores.get('Math')
history_score = student_scores.get('History', 'Not Available')

print(math_score)
print(history_score)
```

* **Explanation:**
 * In this example, we use `get()` to look up the score for 'Math', which gives us 90.
 * When we try to find the score for 'History', which is not present in the dictionary, it returns 'Not Available' instead of throwing an error.

### **`update()` method**
- The `update()` method is used to add new key-value pairs to a dictionary or to change the value of existing keys.
- It’s a quick way to update multiple pieces of information in your dictionary.

##### **Updating multiple items in a dictionary**

```python
student_scores = {'Math': 90, 'English': 85, 'History': 80}

student_scores.update({'Science': 92, 'English': 88, 'History': 85})

print(student_scores)
```

* **Explanation:**
 * Here, `update()` adds a new subject 'Science' with a score of 92, and it also modifies the scores for 'English' and 'History' at the same time.
 * This method is very efficient when you need to make several changes at once, saving you from writing multiple lines of code.

##### **Merging information of Two dictionaries**

```python
employee_details_1 = {
    'Alice': {'department': 'HR', 'position': 'Manager'},
    'Bob': {'department': 'IT', 'position': 'Developer'}}

employee_details_2 = {
    'Charlie': {'department': 'Finance', 'position': 'Analyst'},
    'Bob': {'department': 'IT', 'position': 'Senior Developer'}}

employee_details_1.update(employee_details_2)

employee_details_1
```

* **Explanation:**
 * The `update()` method merges employee_details_2 into employee_details_1.
 * If a key (e.g., 'Bob') exists in both dictionaries, the value in employee_details_1 is updated with the value from employee_details_2.

### **`pop()` method**
- The `pop()` method is used to remove a key-value pair from a dictionary. It also returns the value associated with the key that was removed, allowing you to use or store it if needed.

```python
inventory = {'Apples': 10, 'Bananas': 5, 'Oranges': 8}

removed_quantity = inventory.pop('Bananas')

print(inventory)

print(f"Removed quantity: {removed_quantity}")
```

* **Explanation:**

 * In this example, `pop('Bananas')` removes the 'Bananas' entry from the inventory dictionary and returns its value (5).
 * The dictionary is now updated to exclude 'Bananas', and the removed quantity is stored in the removed_quantity variable for further use.

### **Other methods**
Here is a list of other dictionary methods available for dictionaries:

 * `popitem()`: Removes and returns the last key-value pair from the dictionary. Raises a KeyError if the dictionary is empty.

 * `setdefault()`: Returns the value of a key if it exists; otherwise, inserts the key with a specified default value and returns that value.

 * `clear()`: Removes all items from the dictionary, making it empty.

 * `copy()`: Returns a shallow copy of the dictionary, meaning it creates a new dictionary with the same keys and values.
