# **For Loop**

#### **Overview**
* A For Loop helps you repeat a block of code for each item in a list or a **collection**
* In Python a collection refers to a container that stores multiple items, e.g., List, Tuple, String, etc.
* To use a for loop, you need to specify the collection you want to iterate over and the code you want to run for each item in the collection

#### **Understanding working of a for loop**
* In the below example, we are running a for loop on a list `fruits`
* When the code executes, Python automatically goes through each item in the collection (list here) one by one
* The word `fruit` acts as a **loop variable (iterator)** that takes on the value of each item in the collection during each iteration of the loop
 * Instead of `fruit` you can pick any name of choice _(avoid Keywords like `print, list, range, for`, etc.)_
* The For loop runs as many times as the number of items in the collection
* For each iteration, the code specified in the **code block** executes.
 * Here it is just a `print()` statement

```python
fruits = ['apple', 'banana', 'cherry']
for fruit in fruits:
    print(fruit)
```

* After the loop has executed, the loop variable stores the value of the last item of collection

```python
print(fruit)
```

#### **When to use a For loop**
* For loops are useful when you need to perform the same action on each item in a collection.
* For example, if you have a list of numbers and you want to add 1 to each number and print the outcome, a for loop can help you do that easily.

```python
nums = list(range(10,60,10))
print(nums)
for i in nums:
  print(i+1)
```

#### **Running For loop on other Collections**

# **While Loop**

#### **Overview**
* A while loop is a type of loop that repeatedly executes a block of code as long as a certain condition is true.
* The condition is evaluated before each iteration of the loop.
* The loop will continue to run as long as the condition remains true.

#### **Understanding working of a while loop**
* In the below example, we are running a while loop that counts down from 5 to 1.
* When the code executes, Python checks the condition before each iteration of the loop.
* The condition in this case is count > 0.
* The loop variable count is initialized to 5 before the loop starts.
* As long as count is greater than 0, the code block inside the loop executes.
* For each iteration, the code specified in the code block executes.
 * Here, it prints the value of count and then decrements count by 1.
* The loop stops when the condition count > 0 becomes False.

```python
count = 5
while count > 0:
    print(count)
    count -= 1
```

#### **When to use a while loop**
* Use a while loop when
 * You don't know in advance how many times the loop needs to run
 * You need to repeat an action until a specific condition is met. For example, *if you want to keep asking the user for input until they provide a valid response, a while loop can help you do that easily*

##### Example : Validating password

```python
correct_password = "9090"
user_input = ""

while user_input != correct_password:
    user_input = input("Enter the password: ")

print("Access granted!")
```
