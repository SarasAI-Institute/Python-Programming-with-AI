# **This notebook outlines how Python's `dir()` function works and the different contexts in which it can be used**

## **01 Working**
* The primary purpose of the `dir()` function in Python is to list the attributes and methods of an object.
* If you want to see what functions and variables are available within a specific object, module, or class, you can use the `dir()` function followed by the name of the object.
* A better understanding of the `dir()` function allows us to explore and utilize Python's built-in and external objects more effectively.
* It is useful for developers to inspect objects and understand the available properties and methods, aiding in debugging and code comprehension.


## **02 Use**

#### **Displaying `List` Methods**
* For now ignore items which start and end with double underscores (\_\_xxx\_\_) as they will be explained later in Object Oriented Programming Module
* All `List` methods show at the bottom

```python
dir(list)
```

#### **Displaying `Tuple` Methods**

```python
dir(tuple)
```

#### **Displaying `str` Class Methods**
* `dir('hello')` will also return the same output

```python
dir(str)
```

#### **Using `dir()` with no arguments**

* It appears that running dir() does not return anything meaningful

```python
dir()
```

* We'll declare a few variables and re-run dir()

```python
Name = 'John Smith'
Age = 35
Gender = 'Male'
City = 'Kyoto'
```

* Notice that the above Four variable names now appear in the below `dir()` function output

```python
dir()
```

* Thus without any arguments, the `dir()` function shows all the names of variables that were created (if any) in the current workspace/ session
* **Note :** In the later Modules, when we create user defined functions or import functions from external modules (`math`, `random`, `os`, etc.), `dir()` will display them as well

## **03 Note on the `dir()` function**
* The `dir()` function in Python is a valuable tool that lists the attributes and methods available for an object. It helps you discover the capabilities of various objects.
* In later modules, you'll learn more about using `dir()` to explore libraries, examine user-defined objects, and understand the properties and methods they offer. This aids in debugging and helps you better utilize objects in your programs.
