# **List Comprehension**
This notebook introduces you to List comprehension in Python. You'll learn how to use this powerful feature to create and manipulate lists more efficiently. It's a great way to write cleaner and more readable code while performing tasks like filtering, transforming, and flattening lists. We'll explore various examples to demonstrate its practical applications and advantages.

# **Introduction to List Comprehension**

### **A Simple Example of For Loop**

We have learned how to create a list of numbers using a traditional for loop. Here's a quick refresher
* In this example, we create a list of squares by iterating through numbers from 0 to 4 and appending their squares to the list.

````python
squares = []
for x in range(5):
    squares.append(x**2)
print(squares)
````

### **Mini Challenge - Can you do it in one line?**
Wouldn't it be amazing if you could achieve the same result with just one line of code?
##### **With list comprehension, you can do just that!**

````python
squares = [x**2 for x in range(5)]
print(squares)
````

* __In the above example `[x**2 for x in range(5)]` generates a list of squares directly, making the code shorter and more elegant__
* This is the list comprehension way of writing a for loop to perform the same operation.

### **What is List Comprehension?**
List comprehension is a concise and powerful way to create lists in Python.
* Instead of writing multiple lines of code in the above for loop example, we used a single line that combined the loop and the list creation into one.
* Using a traditional for loop to create a list can sometimes result in more verbose code. List comprehension allows you to achieve the same result with less code, making it easier to understand and maintain.

# **List Comprehension versus normal for loops**

### Differences in writing List Comprehension vs. Normal for loops

#### **List Initialization**
  - **Normal For Loop:**
    ```python
    squares = []
    ```
    You start by explicitly initializing an empty list before entering the loop.

  - **List Comprehension:**
    ```python
    squares = [x**2 for x in range(5)]
    ```
    The list is created directly within the list comprehension syntax, eliminating the need for a separate initialization step.

#### **Appending Elements**
  - **Normal For Loop:**
    ```python
    for x in range(5):
        squares.append(x**2)
    ```
    You use the `.append()` method to add each element to the list within the loop. This requires indentation and a colon at the end of the `for` line.

  - **List Comprehension:**
    ```python
    squares = [x**2 for x in range(5)]
    ```
    Elements are added to the list automatically as part of the comprehension expression, without needing separate lines for appending or indentation.

#### **Syntax and Structure**
  - **Normal For Loop:**
    The code is spread out over multiple lines:
    - **Initialization:** `squares = []`
    - **Loop:** `for x in range(5):`
    - **Action (Indented):** `squares.append(x**2)`

  - **List Comprehension:**
    All operations are combined into a single line enclosed in `[ ]` brackets:
    ```python
    squares = [x**2 for x in range(5)]
    ```
    No need for separate initialization, colons, or indentation.

#### **Memory Efficiency**
  - **Normal For Loop:**
    In the traditional approach, you first create and store the entire list before applying any operations to it. This means the list is fully built and saved in memory, which can be less efficient for very large datasets.

    In this example, the list squares contains the squares of numbers from 1 to 9999. The list is fully constructed and stored in memory before the `sum()` function is applied to it. This means that a large amount of memory is used to hold the list, which can be inefficient, especially if the list is very large.

    ```python
    squares = []
    for x in range(1, 10000):
        squares.append(x**2)
    total = sum(squares)

  - **List Comprehension:**
    With list comprehension, you can directly apply functions to the generated list without creating an intermediate list if it’s not necessary. This helps in reducing memory usage.

    In this example, the `sum()` function operates directly on the list created by the list comprehension. Although the list is created in memory temporarily, it is not assigned to a variable. Once the `sum()` function has used the list, the List is immediately discarded, freeing up the memory. This temporary nature of the list helps in reducing memory usage compared to storing the list in a variable.
    
    ```python
    total = sum([x**2 for x in range(1, 10000)])


#### **Readability and Performance**
- **Readability:**
  - **Normal For Loop:** More verbose but may be clearer for complex operations or when debugging.
  
  - **List Comprehension:** More compact and readable for simple tasks, but can become difficult to read with more complex logic.

- **Performance:**
  - **Normal For Loop:** Generally slower due to the overhead of multiple lines and method calls.

  - **List Comprehension:** Typically faster and more optimized for performance in simple cases.

# **List Comprehension Deep-Dive**

##### **Example 01:** You have a list of numbers and you want to add 5 to each number.

````python
numbers = [1, 2, 3, 4, 5]

print([num + 5 for num in numbers])
````

##### **Example 02:** You have a list of usernames, and you want to filter out usernames that are shorter than 5 characters.

````python
usernames = ["alice", "bob", "charlie", "david", "eve"]
[name for name in usernames if len(name) >= 5]
````

##### **Example 03:** You have a list of temperatures in Celsius and want to convert them to Fahrenheit.

````python
celsius_temps = [0, 20, 56, 100]
fahrenheit_temps = [((9/5) * temp + 32) for temp in celsius_temps]
print(fahrenheit_temps)
````

##### **Example 04:** You have a list of numbers and want to find out which ones are divisible by either 2 or 5.

````python
numbers = [10, 23, 30, 45, 50, 77]

divisible_by_2_or_5 = [num for num in numbers if num % 2 == 0 or num % 5 == 0]
print(divisible_by_2_or_5)
````

##### **Example 05:** You have a list of movie titles and want to find all movies that include the word "Star" in their titles.

````python
movies = ["Star Wars A New Hope", "The Dark Knight", "Star Trek",
          "Inception", "Guardians of the Galaxy", "Star Wars The Last Jedi"]

star_movies = [movie for movie in movies if "Star" in movie]
print(star_movies)
````

##### **Example 06:** You have a list of product prices and want to apply a discount of 10% only to prices greater than $50.

````python
prices = [45, 60, 120, 30, 85]
discounted_prices = [price * 0.9 if price > 50 else price for price in prices]
print(discounted_prices)
````

##### **Example 07:** You want to create a multiplication table for numbers 5 to 9.

 * In this example, we are creating a multiplication table for the numbers 5 through 9.
  * The outer list comprehension for x in range(5, 10) iterates over the numbers 5 to 9.
  * For each number x, the inner list comprehension [x * y for y in range(1, 11)] calculates the products of x with numbers from 1 to 10.
 * The result is a list of lists where each inner list represents a row of the multiplication table for a specific number.

````python
table = [[x * y for y in range(1, 11)] for x in range(5, 10)]
table
````

##### **Example 08:** Imagine you have grocery lists from three different family members. Each list contains items they plan to buy. You want to create a single list that combines all the items from each list to make sure nothing is missed.

````python
grocery_lists = [["milk", "eggs", "bread"],
                ["apples", "bananas", "chicken"],
                ["rice", "beans", "cheese"]]

all_items = [item for sublist in grocery_lists for item in sublist]
print(all_items)
````

 * **Understanding the Structure:**
  * `grocery_lists` is a Nested list that contains several sublists.
Each sublist represents a list of grocery items.

 * **List Comprehension:**
  * `[item for sublist in grocery_lists for item in sublist]` is used to create a flat list from the nested list.

 * **Order of Loops:**
  * First Loop `for sublist in grocery_lists`: This is the outer loop. It goes through each sublist in grocery_lists.
    * On the first iteration, sublist will be `["milk", "eggs", "bread"]`.
    * On the second iteration, sublist will be `["apples", "bananas", "chicken"]`.
    * On the third iteration, sublist will be `["rice", "beans", "cheese"]`.
   * Second Loop `for item in sublist`: This is the inner loop. For each sublist, this loop goes through each item in that sublist.
  
    * For the first sublist, item will be "milk", "eggs", and "bread".
    * For the second sublist, item will be "apples", "bananas", and "chicken".
    * For the third sublist, item will be "rice", "beans", and "cheese".

 * **Flattening the List:**
  * The outer loop picks each sublist one by one.
The inner loop then picks each item from the current sublist.
All these item values are collected into the new list all_items.

 * **Result:**
  * The all_items list contains all the items from all the sublists, but in a single flat list.

##### **Example 09 :** You and a friend have created separate playlists of your favorite movies. Now, you want to discover which movies appear in both playlists to see if you share any favorites.

````python
playlist_friend_1 = ["Inception", "The Matrix", "Interstellar", "The Dark Knight", "Fight Club"]
playlist_friend_2 = ["Fight Club", "Inception", "Pulp Fiction", "The Godfather", "The Matrix"]

common_movies = [movie for movie in playlist_friend_1 if movie in playlist_friend_2]
print(common_movies)
````

##### **Example 10:** You have a nested list where each sub-list represents scores of students in different subjects. To find the top score in Subject 1 (the first subject), you need to compute the maximum score across all students for that subject.

````python
student_scores = [[85, 90, 78], [92, 88, 95], [87, 91, 80]]
top_score_subject_1 = max([scores[0] for scores in student_scores])
print(top_score_subject_1)
````

# **Limitations of Using List Comprehension**

   List comprehensions are a powerful feature in Python, but they have limitations. Here’s an overview of where they might not be the best choice and why normal for loops might be preferable:
 * **Nested Loops**
  * Nested list comprehensions can be hard to understand. Normal for loops offer better clarity for deeply nested structures.
 * **Complex Operations**
  * List comprehensions can become difficult to read and debug when operations are complex. Normal for loops are better for handling complex logic and multiple conditions.
 * **Side Effects**
  * List comprehensions are not ideal for situations where you need to perform actions with side effects, like printing messages or updating external variables.
  * In these cases, a normal for loop is more appropriate because it can handle multiple actions more clearly.
  
  **Scenario:** Suppose you want to print each item in a list and also count how many items have been processed. This involves side effects (printing and counting) that are better managed with a normal for loop.

````python
count = 0
items = ["apple", "banana", "cherry"]

for item in items:
    print(f"Processing {item}")
    count += 1

print(f"Total items processed: {count}")
````
